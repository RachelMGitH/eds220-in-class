```python
# Importing pandas and numpy, always import all packages in a single cell at the top of the notebook
import pandas as pd
import numpy as np
```

```python
# Using the format s = pd.Series(data, index=index)
# The data can be a list, array, dictionary, sinlge number, boolean, or string
# Creating a pandas.Series from a NumPy array

pd.Series(np.arange(3), index = [2023,2024,2025])
```

```python
# Creating a pandas.Series from a list of strings with default index

pd.Series(['EDS 220', 'EDS 222', 'EDS 223', 'EDS 242'])
```

```python
# Creating a pandas.Series from a dictionary

# Construct dictionary
d = {'key_0':2, 'key_1':'3', 'key_2':5}

#Initialize series using a dictionary
pd.Series(d)
```

```python
# Crearing a pandas.Series from a single value

pd.Series(3.0, index = ['A', 'B', 'C'])
```

Simple Operations

```python
# Define a series
s = pd.Series([98,73,65], index=['Andrea', 'Beth', 'Carolina'])

# Divide each element in series by 10
print(s /10, '\n')

# Take the exponential of each element in a series
print(np.exp(s), '\n')

# Original series is unchanged
print(s)
```

```python
# Producing a new pandas.Series with True/False values for a condition

s>70
```

Identifying Missing Values

```python
# Series with NAs in it
s = pd.Series([1,2,np.nan, 4, np.nan])
s
```

```python
# Check if series has NAs
s.hasnans
```

```python
# Checking which elements in the series are NAs
s.isna()
```

```python
# Check in
# Create a pandas.Series named s with four integer values, two of which are -999. 
# The index of this series should be the letters A through D.

s = pd.Series([1,-999,2,-999], index=['A', 'B', 'C', 'D'])
s
```

```python
# Using mask() to replace the -999 values with NA.
?pd.Series.mask
```

```python
s = s.mask(s == -999)
s
```

Data Frames

```python
# Creating a pandas.DataFrame

# Initialize dictionary with columns' data
d = {'col_name_1' : pd.Series(np.arange(3)),
     'col_name_2' : pd.Series([3.1,3.2,3.3])
    }

# Create data frame
df = pd.DataFrame(d)
df
```

```python
# Change index
df.index = ['a','b','c']
df
```

```python
# Check-in

# Update the column names to C1 and C2 via the columns attribute

df.columns = ['C1','C2']
df
```
